# Import Libraries 

```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

# Define Dataset Path

```python
DATA_PATH = "../../images"

categories = ["building", "checker", "grid", "object", "shape"]
```

# Image Loader Function

```python
def load_images_by_prefix(data_path, categories):
    dataset = {cat: [] for cat in categories}
    
    for file in os.listdir(data_path):
        if file.lower().endswith((".jpg", ".png", ".jpeg")):
            for category in categories:
                if file.startswith(category):
                    img_path = os.path.join(data_path, file)
                    img = cv2.imread(img_path)
                    
                    if img is not None:
                        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                        dataset[category].append({
                            "filename": file,
                            "image": img
                        })
    
    return dataset
```

# Load Dataset

```python
dataset = load_images_by_prefix(DATA_PATH, categories)

for category in dataset:
    print(f"{category}: {len(dataset[category])} images")
```

# Display Sample Images

```python
def show_samples(dataset, category, num=3):
    plt.figure(figsize=(15,5))
    
    for i in range(min(num, len(dataset[category]))):
        plt.subplot(1, num, i+1)
        plt.imshow(dataset[category][i]["image"])
        plt.title(dataset[category][i]["filename"])
        plt.axis("off")
    
    plt.suptitle(f"Category: {category}")
    plt.show()

show_samples(dataset, "checker")
```
